# Stets Home

Дипломный проект: требования и прототип фирменного приложения для управления умным домом компании Stets.

Тетрадь фиксирует результаты работы: потребности заказчика и пользователей, границу MVP, модели данных и процессов, прототип в Figma и программу испытаний.


## Описание проекта

К компании RANA.Software обратилась Stets — производитель умных электронных устройств. Устройства Stets уже совместимы со сторонними приложениями умного дома, но этого стало недостаточно: компания разработала программно-аппаратный способ запуска устройств в режиме энергосбережения и готовит к выпуску первую умную лампочку с включением и выключением этого режима. В приложениях других производителей такой функции нет, поэтому Stets заказала собственное приложение **Stets Home**.

Работа состоит из двух частей.

**Часть 1. Требования.** Анализ интервью с представителем Stets и потребностей пользователей. Результат: карта пользовательских историй с границей MVP, контекстная и логическая DFD, логическая ER-диаграмма в нотации Crow’s Foot, словарь данных.

**Часть 2. Прототип.** Сборка интерфейса в Figma на дизайн-системе Stets, программа и методика испытаний (ПМИ), видеопрезентация прототипа по шагам ПМИ.

Итоговый артефакт — прототип первой версии приложения, в котором реализованы функции MVP.

### Контекстная и логическая DFD

Диаграммы хранятся в файле `Диаграммы.drawio`.

**Контекстная DFD.** Внешние сущности: пользователь, устройства умного дома, сервис доставки сообщений (email, SMS). Центральный процесс — приложение «умный дом». Основные потоки: данные регистрации и входа, запросы управления домами, комнатами, устройствами, сценариями и энергосбережением, телеметрия устройств, коды и приглашения через email/SMS.

**Логическая DFD.** Процессы: управление учётной записью, домами, устройствами, комнатами, энергосбережением, сценариями и совместным доступом. Хранилища: D1 «Пользователи», D2 «Дом», D3 «Устройства», D4 «Комнаты».


```python
import re
import xml.etree.ElementTree as ET

drawio = ROOT / 'Диаграммы.drawio'
root = ET.parse(drawio).getroot()

print('Страницы файла Диаграммы.drawio:')
for i, diagram in enumerate(root.findall('diagram'), start=1):
    print(f'  {i}. {diagram.get("name")}')

logical = next(d for d in root.findall('diagram') if d.get('name') == 'DFD логическая')
values = []
for cell in logical.iter('mxCell'):
    raw = cell.get('value') or ''
    text = re.sub(r'<[^>]+>', ' ', raw)
    text = text.replace('&nbsp;', ' ').replace('&#xa;', ' ')
    text = re.sub(r'\s+', ' ', text).strip()
    if text:
        values.append(text)

print()
print('Процессы и хранилища логической DFD:')
seen = set()
for text in values:
    key = text.lower()
    if key in seen:
        continue
    if any(marker in text.lower() for marker in (
        'управление', 'd1', 'd2', 'd3', 'd4', 'пользователь',
        'устройства умного дома', 'сервис доставки'
    )) and 'запрос' not in text.lower() and 'данные' not in text.lower()[:6]:
        seen.add(key)
        print(f'  - {text}')
```

### Логическая ER-диаграмма и словарь данных

ER-диаграмма выполнена в нотации Crow’s Foot и согласована со словарём данных (`Словарь данных_.docx`).

Сущности модели: пользователь, дом, пользователь дома, приглашение, комната, тип комнаты, устройство, тип устройства, сценарий, действия сценария.

Ключевые ограничения предметной области:

- состояние устройства: `on` / `off`;
- заводской идентификатор устройства: `оборудование_id` (QR или ID с коробки);
- роли в доме: владелец, жилец, сосед;
- статус приглашения: ожидает, принято, отклонено;
- сценарий хранит расписание и целевые состояния устройств.


```python
entities = [
    ('Пользователь', 'пользователь_id', 'email, пароль, имя, телефон, код/ссылка восстановления, активный дом_id'),
    ('Дом', 'дом_id', 'название дома'),
    ('Пользователь дома', 'пользователь дома_id', 'пользователь_id, дом_id, роль, статус доступа'),
    ('Приглашение', 'приглашение_id', 'дом_id, приглашающий, email приглашённого, роль, статус'),
    ('Комната', 'комната_id', 'дом_id, тип комнаты_id, название'),
    ('Тип комнаты', 'тип комнаты_id', 'название типа комнаты'),
    ('Устройство', 'устройство_id', 'дом_id, комната_id, тип, оборудование_id, название, состояние, сеть, энергосбережение'),
    ('Тип устройства', 'тип устройства_id', 'название типа устройства'),
    ('Сценарий', 'сценарий_id', 'дом_id, название, время окончания, активен/не активен'),
    ('Действия сценария', 'действия сценария_id', 'сценарий_id, устройство_id, состояние питания, порядок'),
]

print(f'{"Сущность":<22} {"Ключ":<24} Атрибуты')
print('-' * 110)
for name, key, attrs in entities:
    print(f'{name:<22} {key:<24} {attrs}')
print()
print('Сущностей в логической модели:', len(entities))

er = next(d for d in root.findall('diagram') if d.get('name') == 'ER диаграмма')
er_names = []
for cell in er.findall('.//mxCell'):
    parent = cell.get('parent')
    value = cell.get('value') or ''
    if parent == '1' and value and 'html=1' not in (cell.get('style') or ''):
        er_names.append(value)
    elif parent == '1' and 'swimlane' in (cell.get('style') or ''):
        er_names.append(value)

print()
print('Сущности на ER-диаграмме:')
for name in er_names:
    print(f'  - {name}')
print('Совпадение числа сущностей словаря и ER:', len(entities) == len(er_names))
```
